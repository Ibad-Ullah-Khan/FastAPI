## __POST Request__
A POST request is an HTTP method used to send data to a server, commonly to create a new resource.

**For example**, in our Patient Management System, when we want to add a new patient, we use a POST request.

The client sends patient information to the API through the **request body**.

### __Request Body__
A request body is the data that a client sends to the server inside an HTTP request.

For example, suppose we are building a Patient Management System and want to add a new patient.

We need to send patient information to our API:
- Name
- Age
- Gender
- Height
- Weight

This information is sent in the request body.
