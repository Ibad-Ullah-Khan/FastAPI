# Server:
A server is a computer or system that provides data, services, or resources to other computers (called clients) over a network.



```python
"""
You
 ↓
Browser
 ↓
Request → Server
 ↓
Server processes request
 ↓
Response ← Server
 ↓
Browser shows the result
"""
```

| Server                                                                            | Web Server                                                   |
| --------------------------------------------------------------------------------- | ------------------------------------------------------------ |
| A general computer/system that provides services or resources to other computers. | A specific type of server that handles **web requests**.     |
| Can provide many types of services.                                               | Mainly provides **web pages and web content**.               |
| Example: Database server, file server, mail server.                               | Example: Apache, Nginx.                                      |
| Can work with different types of applications and protocols.                      | Mainly uses **HTTP/HTTPS** to communicate with web browsers. |


# HTTP

HTTP stands for HyperText Transfer Protocol.

It is a set of rules that allows a browser (client) and a web server to communicate with each other.

```python
""" 
You open a website
       ↓
Browser sends HTTP Request
       ↓
Web Server receives the request
       ↓
Server processes it
       ↓
Server sends HTTP Response
       ↓
Browser displays the result
"""
```

# Virtual Environment:
A virtual environment is an isolated Python environment created for a specific project.

It allows each project to have its own Python packages and package versions without affecting other projects.

```python
# 1: Creating Virtual Environment:
# First open you project in terminal the run the following commond in the termial

"python -m venv Project1" # Run this without the double qoutes, it will create a envirnoment of name : Project1

# 2:Activate Virtual Environment
".\Project1\Scripts\Activate.ps1"
# But it can also be activated like this
".\Project1\Scripts\activate"

# 3: Switch to Another Virtual Environment
# First deactive the current virtual environment
"deactivate"

# Then
".\another_env\Scripts\activate"

# 4: Close / Deactivate Virtual Environment
"deactivate"
```

# Status Code

An HTTP status code is a number that tells the client/browser what happened with the request.

| Status Code | Meaning                    | Simple Explanation                 |
| ----------- | -------------------------- | ---------------------------------- |
| **200**     | OK                         | Request was successful             |
| **201**     | Created                    | Something was successfully created |
| **301**     | Moved Permanently          | Resource has permanently moved     |
| **302**     | Found / Temporary Redirect | Redirected to another URL          |
| **400**     | Bad Request                | Request was incorrect              |
| **401**     | Unauthorized               | Authentication is required         |
| **403**     | Forbidden                  | Access is not allowed              |
| **404**     | Not Found                  | Requested resource doesn't exist   |
| **500**     | Internal Server Error      | Something went wrong on the server |


# HTTP Method:
HTTP methods are instructions that tell a web server what you want to do with a resource.

When your browser or application communicates with a server, it sends an HTTP request. The HTTP method describes the action you want to perform.

| HTTP Method | Simple Meaning             | Example                   |
| ----------- | -------------------------- | ------------------------- |
| **GET**     | Get/read something         | Open a web page           |
| **POST**    | Send/create something      | Submit a login form       |
| **PUT**     | Update something           | Update your profile       |
| **PATCH**   | Partially update something | Change only your username |
| **DELETE**  | Delete something           | Delete a post             |
