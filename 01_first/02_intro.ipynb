## __API__
APIs are mechanisms that enable two software components—such as the frontend and backend of an application—to communicate with each other using a defined set of rules, protocols, and data formats
- Rules =>	What each side is allowed or expected to do
- Protocols	=> Rules for how information travels
- Data formats => How information is organized

## __FastAPI__
Is a modern, high performance web framework for building APIs with python
- It is build on __Starlette__ and __Pydantic__. Let's see what these both libraries works

#### __Starlette__
Starlette manages how your API receives requests and sends back responses
#### __Pydantic__
Pydantic is used to check if the data coming into your API is correct and in the right format

### __Why FastAPI is Fast to Run__
##### __1. Asynchronous programming (async/await)__
FastAPI supports asynchronous programming.

Imagine:
- A customer requests some data from a database.
- The database takes 2 seconds to respond.
- During those 2 seconds, FastAPI can work on other requests instead of just waiting.
This helps handle multiple requests efficiently.
##### __2. Starlette__
FastAPI uses a library called Starlette internally.

Starlette handles things like:
- HTTP requests and responses.
- Routing.
- Asynchronous communication.
It helps FastAPI work efficiently.
##### __3. Pydantic__
FastAPI uses Pydantic for data validation.
For example, suppose your API expects a user's age to be a number.
If someone sends "twenty" instead of 20, Pydantic can detect the invalid data.
It makes data handling easier and efficient.

__Asynchronous programming__ means a program can start a task, and while that task is waiting, it can work on other tasks instead of sitting idle.
