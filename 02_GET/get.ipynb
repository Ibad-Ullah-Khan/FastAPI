# GET Method in FastAPI

### 1. What is GET?

GET is an HTTP method used to **retrieve (fetch) data from a server**.

For example:
- Get all patient records.
- Get information about a particular student.
- View user details.

### 2. Creating a GET Endpoint

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Hello World"}
```

### 3. Explanation

- `app = FastAPI()` → Creates a FastAPI application.
- `@app.get("/")` → Creates a GET endpoint at the root URL.
- `def home()` → Function that handles the request.
- `return` → Sends the response back to the client.

### 4. Example Endpoints

| Endpoint | Purpose |
|---|---|
| `/` | Home endpoint |
| `/about` | Get information about the API |
| `/view` | Retrieve patient records |
| `/patients` | Get all patients |

### 5. Important Points

- GET is used to retrieve data.
- GET requests should not modify or delete server data.
- GET responses commonly return JSON data.
- We can create multiple GET endpoints in one FastAPI application.

**Syntax:**

```python
@app.get("/endpoint")
def function_name():
    return {"message": "Response"}
```
