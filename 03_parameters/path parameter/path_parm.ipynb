### __1. What is a Path Parameter?__
A Path Parameter is a variable that we include directly in the URL to identify a specific resource or retrieve specific data.

For example, suppose we have these URLs:
- /patients/1
- /patients/2
- /patients/3

Here, 1, 2, and 3 are path parameters. They represent different patient IDs.

### __1. What is Path()?__
Path() is a FastAPI function used to add extra validation, restrictions, and documentation to path parameters.

For example, suppose you want a patient's ID to be greater than zero.

You can use Path() to enforce this condition.

```python
from fastapi import FastAPI, Path

app = FastAPI()

@app.get("/patients/{patient_id}")
def get_patient(patient_id: int = Path(..., gt=0)):
    return {"patient_id": patient_id}

# Those three dots ... in Path(...) represent Ellipsis in Python.
# In FastAPI, Ellipsis is used to indicate that a parameter is required.
```

### __Status Code__

An HTTP status code is a number that tells the client/browser what happened with the request.

| Status Code | Meaning                    | Simple Explanation                 |
| ----------- | -------------------------- | ---------------------------------- |
| **200**     | OK                         | Request was successful             |
| **201**     | Created                    | Something was successfully created |
| **301**     | Moved Permanently          | Resource has permanently moved     |
| **302**     | Found / Temporary Redirect | Redirected to another URL          |
| **400**     | Bad Request                | Request was incorrect              |
| **401**     | Unauthorized               | Authentication is required         |
| **403**     | Forbidden                  | Access is not allowed              |
| **404**     | Not Found                  | Requested resource doesn't exist   |
| **500**     | Internal Server Error      | Something went wrong on the server |


| Status Code | Category | Meaning |
|---|---|---|
| **2xx** | Success | The request was successfully received and processed. |
| **3xx** | Redirection | Further action needs to be taken (e.g., redirect). |
| **4xx** | Client Error | Something is wrong with the request from the client. |
| **5xx** | Server Error | Something went wrong on the server side. |

### __What is HTTPException?__
HTTPException is used to return an HTTP error response when something goes wrong while processing a request.

For example, suppose you have a patient management system.

A user requests patient ID P999, but that patient doesn't exist in your database.

Instead of returning incorrect information, you can raise an HTTPException with status code 404 (Not Found).

```python
from fastapi import FastAPI, HTTPException

app = FastAPI()

@app.get("/patients/{patient_id}")
def view_patient(patient_id: str):

    if patient_id != "P001":
        raise HTTPException(
            status_code=404,
            detail="Patient not found"
        )

    return {"patient_id": patient_id}
```
